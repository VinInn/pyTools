```python
import os
import re
import matplotlib
import matplotlib.pyplot as plt
import glob
import math
import numpy as np
import jax as jp
import jax.numpy as jnp
from jax import jit
```

```python
x = np.array([884279719003555.0,1.2,2.4,2.1,2.0], dtype=np.float32);
y = np.array([884279719003555.0,1.2,2.4,2.1,2.0], dtype=np.float32);
```

```python
def foo(x,y) :
    return x*x - y*y
def foo2(x,y) :
    x2 = x*x
    y2 = y*y
    return x2 - y2
```

```python
z=foo(x,y)
print(z)
```

```python
xj = jnp.array(x) # [884279719003555.0,1.2,2.4,2.1,2.0]);
yj = jnp.array(y) # [884279719003555.0,1.2,2.4,2.1,2.0]);
fooj = jit(foo)
foo2j = jit(foo2)

zj=fooj(xj,yj)
print(zj)
zj=foo2j(xj,yj)
print(zj)
```

```python
def fma(a,b,c) :
    return a*b+c

fmaj = jit(fma, inline=True)

def augmentedMultiply(b,c) :
   w = b*c 
   e = fmaj(b,c,-w)
   return [w,e]


print(augmentedMultiply(x,y))

am = jit(augmentedMultiply, inline=True)
print(am(xj,yj))
print(fmaj(xj,yj,-xj*yj))
print(fma(xj,yj,-xj*yj))
# print(np.fma(x,y,-x*y))
```

```python

def kahanDet(a, b, c, d) :  # a*d-b*c 
   w = b*c 
   e = w -b*c # error on w: (w,-e) = augmentedMultiply(b,c)
   f = a*d - w 
   return (f+e) 

kdj = jit(kahanDet)
```

```python
zj=kdj(xj,yj,yj,xj)
print(zj)
```
